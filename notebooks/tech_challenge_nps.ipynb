## 1.1 Problema do negócio
O e-commerce brasileiro cresceu nos últimos anos. Segundo a ABComm/Abiacom, o faturamento passou de R\$ 126,45 bilhões em 2020 para R\$ 235,5 bilhões em 2025, um aumento nominal de aproximadamente 86%, sem considerar a inflação. Nesse contexto de crescimento, a empresa passou a lidar com um volume maior de pedidos, entregas e interações com clientes. Embora esse aumento tenha trazido ganhos de escala, também surgiram desafios na experiência do cliente, percebidos pela variação do NPS entre diferentes perfis, o NPS só chega depois da jornada, e a empresa quer agir antes da pesquisa. O problema é entender por que clientes com experiências operacionais aparentemente semelhantes avaliam a empresa de formas tão diferentes e quais fatores podem estar relacionados a essas avaliações.

## 1.2 Importância do NPS
O NPS é importante para um e-commerce porque mostra se os clientes estão satisfeitos com a experiência de compra e se recomendariam a loja para outras pessoas. As respostas também ajudam a perceber o que está funcionando bem e o que pode melhorar, como o atendimento, a entrega ou os produtos. Assim, a empresa entende melhor a opinião de quem compra, em vez de depender apenas da própria percepção. O NPS pode ajudar a entender a satisfação, mas não mede sozinho toda a experiência.

## 1.3 Áreas beneficiadas pela análise
Os dados podem ajudar várias áreas da empresa a entender o que está afetando a satisfação dos clientes:
*   Logística: acompanhar atrasos, tempo de entrega, valor do frete e tentativas de entrega. Se esses fatores aparecerem associados a notas menores, a equipe pode revisar transportadoras, prazos e comunicação sobre o pedido.
*   Atendimento: analisar contatos, reclamações e tempo de resolução. Isso pode
indicar problemas recorrentes e oportunidades para agilizar o suporte ou corrigir as causas que levam os clientes a entrar em contato.
*   Pricing e área comercial: avaliar como descontos, valor do pedido e condições de pagamento se relacionam com a satisfação. Assim, a empresa pode verificar se suas ofertas são atrativas e se as condições de compra são claras.
*   Experiência do cliente e marketing: acompanhar as avaliações e identificar perfis ou etapas da jornada que precisam de atenção. Isso pode ajudar a planejar ações de relacionamento e retenção.
*   Gestão: reunir os resultados para decidir quais problemas devem ser priorizados e acompanhar se as mudanças melhoraram a experiência.

## 1.4 Reflexão: como o NPS pode impactar o negócio
O NPS ajuda a entender como os clientes avaliam a experiência com a empresa e se estariam dispostos a recomendá-la. Ele não garante que o cliente vá comprar novamente ou indicar a loja, mas pode mostrar tendências e ajudar a identificar o que merece mais atenção.
Na recompra, por exemplo, é possível comparar as avaliações com o histórico de
compras para verificar se os clientes que dão notas melhores costumam voltar e se experiências negativas estão relacionadas à perda de clientes. No boca a boca, clientes satisfeitos podem recomendar a loja, enquanto aqueles que tiveram uma experiência ruim podem compartilhar o ocorrido, inclusive em avaliações na internet.

Uma boa experiência também pode ajudar a empresa a manter clientes e atrair novos compradores. Isso pode fortalecer sua posição no mercado, mas o NPS, sozinho, não mostra se a participação da empresa aumentou. Para avaliar isso, também é preciso acompanhar as vendas, o número de clientes e o desempenho dos concorrentes.

## 1.5 Indicadores de mercado que podem complementar a análise
Para complementar a análise, seria possível comparar o NPS da empresa com o de outros e-commerces do mesmo setor e observar os prazos de entrega prometidos e cumpridos pelos concorrentes. Também poderiam ser analisados os preços, o frete e as avaliações e reclamações publicadas por clientes dessas lojas. Esses dados ajudariam a entender se os pontos observados na empresa também aparecem no mercado e como a experiência oferecida se compara à dos concorrentes.



## 2.1 Variável escolhida
A variável escolhida para representar a avaliação do cliente é nps_score. Ela registra a nota dada pelo cliente sobre a possibilidade de recomendar a empresa.
Por que essa variável foi escolhida?
Escolhi essa variável porque a análise busca entender por que as avaliações variam entre os clientes e quais fatores podem estar relacionados a notas mais altas ou mais baixas.

## 2.2 Em que momento ela é coletada?
O nps_score é coletado após a experiência de compra.

## 2.3 Quais são os riscos de usar essa variável?
A nota individual não deve ser confundida com o NPS geral da empresa, que é calculado a partir das respostas de vários clientes. Com 4,4% de promotores e 84,36% de detratores, ele é um resultado extremo. Além disso, a nota sozinha não explica o motivo da avaliação.

## 2.4 Definição do alvo
Neste trabalho, o alvo é a nota de NPS dada pelo cliente. Os dados apresentam a nota do NPS em números decimais, para uma analise mais simples em grupos maiores o NPS será dividido em três categorias: detrator, neutro e promotor.

Será feita uma classificação das notas do NPS em três categorias:
*   Detratores (notas de 0 a 6): clientes insatisfeitos, que podem compartilhar uma experiência negativa.
*   Neutros ou passivos (notas 7 e 8): clientes que tiveram uma experiência razoável, mas podem escolher outra empresa se encontrarem uma opção melhor.
*   Promotores (notas 9 e 10): clientes que tiveram uma experiência positiva e têm maior disposição para recomendar a empresa.




```python
def classificar_nps(linha):
    nota = linha["nps_score"]

    if nota < 7:
        return "Detrator"
    elif nota < 9:
        return "Neutro"
    else:
        return "Promotor"

df["nps_categoria"] = df.apply(classificar_nps, axis=1)
```

A base está concentrada em detratores: eles representam 84,36% dos clientes, enquanto os promotores correspondem a 4,4%. Essa diferença entre os tamanhos dos grupos deve ser considerada na leitura dos gráficos e das comparações.

### Variáveis Financeiras

```python
variaveis_financeiras = [
    'order_value',
    'discount_value',
    'freight_value'
]

df[variaveis_financeiras].describe().round(2)
```

```python
# Comparando as variáveis financeiras por categoria de NPS

analise_nps_financeiro = df.groupby('nps_categoria')[
    variaveis_financeiras
].agg(['mean', 'median', 'count']).round(2)

analise_nps_financeiro
```

Irei utilizar a correlação de Spearman para investigar se existe uma associação monotônica entre os valores financeiros e a nota individual de satisfação.
A correlação de Spearman varia de -1 a 1: próximo de +1	associação positiva, próximo de 0	ausência de associação monotônica relevante e próximo de -1	associação negativa.

```python
# Correlação de Spearman

correlacoes = df[
    variaveis_financeiras + ['nps_score']
].corr(method='spearman')

correlacoes['nps_score'].sort_values(ascending=False)
```

Na análise das variáveis financeiras, observa-se que os promotores apresentam, em média, maior valor de pedido, maiores descontos e menor valor de frete em comparação aos detratores. Entretanto, a análise de correlação de Spearman indica que essas variáveis apresentam associações monotônicas muito fracas com a nota de satisfação. Dessa forma, embora existam diferenças nas médias entre as categorias de clientes, os resultados não indicam uma associação monotônica relevante entre as variáveis financeiras analisadas e a nota de satisfação.

```python
# Distribuição das categorias de NPS por faixa de atraso na entrega (%)

faixas = [0, 1, 3, 5, float("inf")]
nomes = ["0 dias", "1–2 dias", "3–4 dias", "5 dias ou mais"]

df["faixa_atraso"] = pd.cut(
    df["delivery_delay_days"],
    bins=faixas,
    labels=nomes,
    right=False
)

# Quantidade de clientes por faixa e categoria NPS
quantidades = pd.crosstab(
    df["faixa_atraso"],
    df["nps_categoria"]
)

# Percentual de cada categoria NPS
percentuais = pd.crosstab(
    df["faixa_atraso"],
    df["nps_categoria"],
    normalize="index"
).mul(100).round(1)

# Acrescentar o total de observações
percentuais["Total de clientes"] = quantidades.sum(axis=1)

percentuais
```

```python
# Compara a proporção de detratores, neutros e promotores em cada faixa de atraso na entrega

tabela_atraso = pd.crosstab(
    df["faixa_atraso"],
    df["nps_categoria"],
    normalize="index"
).mul(100).reindex(
    index=nomes,
    columns=["Detrator", "Neutro", "Promotor"]
)

ax = tabela_atraso.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5),
    color=["#d9534f", "#f0ad4e", "#5cb85c"]
)

ax.set_title("Categorias de NPS por faixa de atraso na entrega")
ax.set_xlabel("Atraso na entrega")
ax.set_ylabel("Clientes (%)")
ax.legend(title="Categoria de NPS", loc="lower right")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Distribuição das categorias de NPS por faixa de reclamações (%)

faixas_reclamacoes = [-1, 0, 2, 4, 6, float("inf")]
nomes_reclamacoes = ["0", "1–2", "3–4", "5–6", "7 ou mais"]

df["faixa_reclamacoes"] = pd.cut(
    df["complaints_count"],
    bins=faixas_reclamacoes,
    labels=nomes_reclamacoes
)

# Quantidade de clientes por faixa e categoria NPS
quantidades = pd.crosstab(
    df["faixa_reclamacoes"],
    df["nps_categoria"]
)

# Percentual de cada categoria NPS
percentuais = pd.crosstab(
    df["faixa_reclamacoes"],
    df["nps_categoria"],
    normalize="index"
).mul(100).round(1)

# Acrescentar o total de observações
percentuais["Total de clientes"] = quantidades.sum(axis=1)

percentuais
```

```python
# Compara a proporção de detratores, neutros e promotores em cada faixa de reclamações

tabela_reclamacoes = pd.crosstab(
    df["faixa_reclamacoes"],
    df["nps_categoria"],
    normalize="index"
).mul(100).reindex(
    index=nomes_reclamacoes,
    columns=["Detrator", "Neutro", "Promotor"]
)

ax = tabela_reclamacoes.plot(
    kind="bar",
    stacked=True,
    figsize=(9, 5),
    color=["#d9534f", "#f0ad4e", "#5cb85c"]
)

ax.set_title("Categorias de NPS por faixa de reclamações")
ax.set_xlabel("Número de reclamações")
ax.set_ylabel("Clientes (%)")
ax.legend(title="Categoria de NPS", loc="lower right")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
pd.crosstab(
    df["nps_categoria"],
    df["repeat_purchase_30d"],
    normalize="index"
).mul(100).round(1).rename(columns={0: "Sem recompra (%)", 1: "Com recompra (%)"})
```

```python
revisao_recompra = df.groupby("nps_categoria")["repeat_purchase_30d"].agg(
    quantidade_clientes="count",
    clientes_com_recompra="sum",
    percentual_com_recompra=lambda valores: valores.mean() * 100
)

print(revisao_recompra.round(1))
```

## 3.1 Fatores críticos para a satisfação
Os fatores mais críticos para a satisfação é o atraso na entrega, o número de reclamações e o tempo de resolução do problemas

## 3.2 Características dos clientes detratores
Os detratores aparecem principalmente entre clientes que tiveram atrasos na entrega e fizeram mais reclamações.

## 3.3 Pontos de ruptura na experiência do cliente
Os resultados indicam possíveis pontos de ruptura, principalmente no atraso na entrega e no número de reclamações. A proporção de detratores aumenta bastante conforme essas duas situações se agravam:
*   Atraso: passa de 51,6% entre clientes sem atraso para 81,1% com 1–2 dias e 97,6% com 3–4 dias. A partir de 5 dias, todos os clientes da faixa são detratores.
*   Reclamações: entre clientes sem reclamações, 17,4% são detratores. Com 1–2 reclamações, esse percentual sobe para 54,6%; com 3–4, chega a 86,1%.
*   Tempo de resolução: a proporção de detratores também cresce, de 75,0% nos casos resolvidos no mesmo dia para 90,4% nos que levam 9 dias ou mais. A mudança é mais gradual.

## 3.4 Tipo de cliente para NPS mais alto ou mais baixo
Pelos dados analisados, os clientes que tendem a dar notas mais altas são os que não tiveram atraso na entrega, não registraram reclamações e tiveram seus problemas resolvidos rapidamente. Os detratores aparecem frequentemente entre os clientes que enfrentam atrasos, fizeram mais reclamações ou esperaram mais pela solução.
